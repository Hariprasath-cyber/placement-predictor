```python
import pandas as pd
import numpy as np
import joblib
import shap
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')
import warnings
warnings.filterwarnings('ignore')

# Load data
X_train = pd.read_csv('../data/X_train.csv')
X_test  = pd.read_csv('../data/X_test.csv')
y_test  = pd.read_csv('../data/y_test.csv').squeeze()

# Load RF model specifically for SHAP
model         = joblib.load('../models/rf_model.pkl')
feature_names = joblib.load('../models/feature_names.pkl')
numeric_cols  = joblib.load('../models/numeric_cols.pkl')
scaler        = joblib.load('../models/scaler.pkl')

print("All artifacts loaded.")
print("Feature names:", feature_names)
```

```python
print("Initializing SHAP TreeExplainer...")
explainer   = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test)

print("Raw shap_values type :", type(shap_values))
print("Raw shap_values shape:", np.array(shap_values).shape)

# Handle all possible output formats
if isinstance(shap_values, list):
    # Old SHAP format — list of 2 arrays, take index 1 (Placed class)
    sv = shap_values[1]
    print("Format: list → using index [1] for Placed class")
elif len(np.array(shap_values).shape) == 3:
    # New SHAP format — shape (n_samples, n_features, n_classes)
    sv = np.array(shap_values)[:, :, 1]
    print("Format: 3D array → using [:,:,1] for Placed class")
else:
    # Already 2D — shape (n_samples, n_features)
    sv = shap_values
    print("Format: 2D array → using as-is")

print("Final sv shape:", sv.shape)
print("Explainer ready.")
```

```python
plt.figure(figsize=(10, 7))
shap.summary_plot(
    sv, X_test,
    plot_type='bar',
    feature_names=feature_names,
    show=False,
    color='#3498db'
)
plt.title('Global Feature Importance (SHAP)', fontsize=14)
plt.tight_layout()
plt.savefig('../data/plot_shap_global.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved → data/plot_shap_global.png")
```

```python
plt.figure(figsize=(10, 7))
shap.summary_plot(
    sv, X_test,
    feature_names=feature_names,
    show=False
)
plt.title('SHAP Beeswarm — Feature Impact Direction', fontsize=14)
plt.tight_layout()
plt.savefig('../data/plot_shap_beeswarm.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved → data/plot_shap_beeswarm.png")
print("\nHow to read this:")
print("  Red  = high feature value")
print("  Blue = low feature value")
print("  Right of 0 = pushes toward Placed")
print("  Left  of 0 = pushes toward Not Placed")
```

```python
def explain_student(student_idx, X_test, explainer, sv, model, feature_names):
    student_data = X_test.iloc[[student_idx]]
    prob         = model.predict_proba(student_data)[0][1]
    actual       = y_test.iloc[student_idx]

    print(f"Student index    : {student_idx}")
    print(f"Actual status    : {'Placed' if actual == 1 else 'Not Placed'}")
    print(f"Predicted prob   : {prob:.1%}")

    # Fix — handle all SHAP output formats
    if isinstance(explainer.expected_value, (list, np.ndarray)):
        base_value = explainer.expected_value[1]  # class 1 = Placed
    else:
        base_value = explainer.expected_value

    # Fix — handle shape (22, 2) format
    shap_vals = sv[student_idx]
    if len(shap_vals.shape) == 2:
        shap_vals = shap_vals[:, 1]  # take Placed class column

    explanation = shap.Explanation(
        values        = shap_vals,
        base_values   = base_value,
        data          = student_data.values[0],
        feature_names = feature_names
    )

    plt.figure(figsize=(12, 6))
    shap.plots.waterfall(explanation, show=False)
    plt.title(f'Student {student_idx} — Placement Probability: {prob:.1%}',
              fontsize=13)
    plt.tight_layout()
    plt.savefig(f'../data/plot_waterfall_student_{student_idx}.png',
                dpi=150, bbox_inches='tight')
    plt.show()

    return prob, shap_vals

# Explain a placed student
print("=" * 50)
print("EXAMPLE 1 — High probability student")
print("=" * 50)
placed_idx = y_test[y_test == 1].index[0] - y_test.index[0]
prob1, sv1 = explain_student(placed_idx, X_test, explainer, sv, model, feature_names)

# Explain a not-placed student
print("\n" + "=" * 50)
print("EXAMPLE 2 — Low probability student")
print("=" * 50)
not_placed_idx = y_test[y_test == 0].index[0] - y_test.index[0]
prob2, sv2     = explain_student(not_placed_idx, X_test, explainer, sv, model, feature_names)
```

```python
# Load original unscaled test data for display purposes
X_test_original = pd.read_csv('../data/X_test.csv')

# Inverse transform only numeric columns to get real values back
numeric_cols = joblib.load('../models/numeric_cols.pkl')
X_test_display = X_test_original.copy()
X_test_display[numeric_cols] = scaler.inverse_transform(
    X_test_original[numeric_cols]
)

print("Original (display) values sample:")
print(X_test_display[numeric_cols].iloc[0].round(2).to_string())
print("\nScaled (model input) values sample:")
print(X_test_original[numeric_cols].iloc[0].round(2).to_string())
```

```python
def analyze_skill_gap(student_idx, model, scaler,
                      feature_names, numeric_cols,
                      X_test_scaled, X_test_display):
    """
    student_idx    : index of student in test set
    X_test_scaled  : scaled data (what model uses for prediction)
    X_test_display : original unscaled data (what we show to student)
    """

    improvement_targets = {
        'cgpa'                      : 8.5,
        'coding_skill_rating'       : 4.5,
        'aptitude_skill_rating'     : 4.5,
        'communication_skill_rating': 4.0,
        'attendance_percentage'     : 85.0,
        'internships_completed'     : 3.0,
        'projects_completed'        : 4.0,
        'backlogs'                  : 0.0,
        'certifications_count'      : 4.0,
        'hackathons_participated'   : 3.0,
        'study_hours_per_day'       : 6.0,
        'twelfth_percentage'        : 80.0,
        'tenth_percentage'          : 80.0,
    }

    # Current student — scaled for model, original for display
    student_scaled  = X_test_scaled.iloc[[student_idx]].copy()
    student_display = X_test_display.iloc[[student_idx]].copy()

    current_prob = model.predict_proba(student_scaled)[0][1]

    recommendations = []

    for feature, target in improvement_targets.items():
        if feature not in feature_names:
            continue

        # Display current value — original unscaled
        current_display_val = student_display[feature].values[0]

        needs_improvement = (
            current_display_val > target if feature == 'backlogs'
            else current_display_val < target
        )

        if needs_improvement:
            # Build improved version using DISPLAY values
            improved_display = student_display.copy()
            improved_display[feature] = target

            # Scale the improved version for model prediction
            improved_scaled = student_scaled.copy()
            improved_scaled[numeric_cols] = scaler.transform(
                improved_display[numeric_cols]
            )

            new_prob = model.predict_proba(improved_scaled)[0][1]
            impact   = (new_prob - current_prob) * 100

            if impact > 0.5:
                recommendations.append({
                    'feature'      : feature,
                    'current_value': round(float(current_display_val), 2),
                    'target_value' : target,
                    'current_prob' : round(current_prob * 100, 1),
                    'new_prob'     : round(new_prob * 100, 1),
                    'impact'       : round(impact, 1)
                })

    recommendations.sort(key=lambda x: x['impact'], reverse=True)
    return current_prob, recommendations


# ── Test on a not-placed student ──────────────────────────────
not_placed_idx = y_test[y_test == 0].index[0] - y_test.index[0]

current_prob, recs = analyze_skill_gap(
    student_idx    = not_placed_idx,
    model          = model,
    scaler         = scaler,
    feature_names  = feature_names,
    numeric_cols   = numeric_cols,
    X_test_scaled  = X_test,
    X_test_display = X_test_display
)

# Replace the print loop at the bottom of Cell 7 with this:
print(f"Current placement probability : {current_prob:.1%}")
print(f"\nTop recommendations to improve placement chances:")
print("-" * 65)

for i, r in enumerate(recs[:5], 1):
    direction    = "Reduce" if r['feature'] == 'backlogs' else "Improve"
    
    # Fix display of -0.0 here directly
    current_val  = abs(r['current_value'])
    # Format nicely — remove unnecessary decimals
    current_disp = int(current_val) if current_val == int(current_val) else current_val
    target_disp  = int(r['target_value']) if r['target_value'] == int(r['target_value']) else r['target_value']

    print(f"{i}. {direction} {r['feature']}")
    print(f"   Current : {current_disp}")
    print(f"   Target  : {target_disp}")
    print(f"   Probability : {r['current_prob']}% → {r['new_prob']}%  "
          f"(+{r['impact']}%)")
    print()
```

```python
joblib.dump(explainer,       '../models/shap_explainer.pkl')
joblib.dump(improvement_targets := {
    'cgpa'                      : 8.5,
    'coding_skill_rating'       : 4.5,
    'aptitude_skill_rating'     : 4.5,
    'communication_skill_rating': 4.0,
    'attendance_percentage'     : 85.0,
    'internships_completed'     : 3.0,
    'projects_completed'        : 4.0,
    'backlogs'                  : 0.0,
    'certifications_count'      : 4.0,
    'hackathons_participated'   : 3.0,
    'study_hours_per_day'       : 6.0,
    'twelfth_percentage'        : 80.0,
    'tenth_percentage'          : 80.0,
}, '../models/improvement_targets.pkl')

print("Saved:")
print("  models/shap_explainer.pkl")
print("  models/improvement_targets.pkl")
print("\nAll notebooks complete.")
print("Ready for → app.py")
```
