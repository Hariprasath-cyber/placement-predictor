```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)
print("Libraries loaded successfully.")
```

```python
df_features = pd.read_csv('../data/indian_engineering_student_placement.csv')
df_target   = pd.read_csv('../data/target.csv')

print("Features shape:", df_features.shape)
print("Target shape  :", df_target.shape)
```

```python
df = pd.merge(df_features, df_target[['Student_ID', 'placement_status']],
              on='Student_ID', how='inner')

# Drop Student_ID — it's just an identifier, not a feature
df.drop('Student_ID', axis=1, inplace=True)

print("Merged shape:", df.shape)
print("Columns:", list(df.columns))
df.head()
```

```python
print("=== Data Types ===")
print(df.dtypes)

print("\n=== Missing Values ===")
missing = df.isnull().sum()
if missing.sum() == 0:
    print("No missing values. Dataset is clean.")
else:
    print(missing[missing > 0])
```

```python
print("\nTarget distribution (placement_status):")
print(df['placement_status'].value_counts())

print("\nPlacement rate:")
print((df['placement_status'].value_counts(normalize=True) * 100).round(2))
```

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(x='placement_status', data=df,
                   palette={'Placed':'#2ecc71', 'Not Placed':'#e74c3c'})
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=12, fontweight='bold')
plt.title('Placement Status Distribution', fontsize=14)
plt.xlabel('')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('../data/plot1_target_distribution.png', dpi=150)
plt.show()
```

```python
# Manually defined based on your dataset structure
numeric_cols = [
    'cgpa', 'tenth_percentage', 'twelfth_percentage',
    'study_hours_per_day', 'attendance_percentage',
    'projects_completed', 'internships_completed',
    'coding_skill_rating', 'communication_skill_rating',
    'aptitude_skill_rating', 'hackathons_participated',
    'certifications_count', 'sleep_hours',
    'stress_level', 'backlogs'
]

categorical_cols = [
    'gender', 'branch', 'part_time_job',
    'family_income_level', 'city_tier',
    'internet_access', 'extracurricular_involvement'
]

print("Numeric features   :", len(numeric_cols))
print("Categorical features:", len(categorical_cols))
```

```python
print("=== Average values — Placed vs Not Placed ===\n")
summary = df.groupby('placement_status')[numeric_cols].mean().round(2)
print(summary.T.to_string())
```

```python
top_numeric = ['cgpa', 'aptitude_skill_rating', 'coding_skill_rating',
               'attendance_percentage', 'internships_completed',
               'communication_skill_rating']

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for i, col in enumerate(top_numeric):
    sns.boxplot(x='placement_status', y=col, data=df,
                ax=axes[i],
                palette={'Placed':'#2ecc71', 'Not Placed':'#e74c3c'})
    axes[i].set_title(f'{col} vs Placement', fontsize=12)
    axes[i].set_xlabel('')

plt.suptitle('Key Numeric Features vs Placement Status', fontsize=14, y=1.01)
plt.tight_layout()
plt.savefig('../data/plot2_numeric_vs_placement.png', dpi=150)
plt.show()
```

```python
remaining_numeric = ['tenth_percentage', 'twelfth_percentage',
                     'study_hours_per_day', 'projects_completed',
                     'hackathons_participated', 'certifications_count',
                     'sleep_hours', 'stress_level', 'backlogs']

fig, axes = plt.subplots(3, 3, figsize=(16, 12))
axes = axes.flatten()

for i, col in enumerate(remaining_numeric):
    sns.boxplot(x='placement_status', y=col, data=df,
                ax=axes[i],
                palette={'Placed':'#2ecc71', 'Not Placed':'#e74c3c'})
    axes[i].set_title(f'{col} vs Placement', fontsize=11)
    axes[i].set_xlabel('')

plt.tight_layout()
plt.savefig('../data/plot3_remaining_numeric.png', dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(3, 3, figsize=(18, 14))
axes = axes.flatten()

for i, col in enumerate(categorical_cols):
    sns.countplot(x=col, hue='placement_status', data=df,
                  ax=axes[i],
                  palette={'Placed':'#2ecc71', 'Not Placed':'#e74c3c'})
    axes[i].set_title(f'{col} vs Placement', fontsize=11)
    axes[i].set_xlabel('')
    axes[i].tick_params(axis='x', rotation=30)
    axes[i].legend(title='', fontsize=8)

# Hide unused subplots
for j in range(len(categorical_cols), len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Categorical Features vs Placement Status', fontsize=14, y=1.01)
plt.tight_layout()
plt.savefig('../data/plot4_categorical_vs_placement.png', dpi=150)
plt.show()
```

```python
temp = df[numeric_cols].copy()
temp['placed'] = (df['placement_status'] == 'Placed').astype(int)

plt.figure(figsize=(14, 10))
sns.heatmap(temp.corr(), annot=True, fmt='.2f',
            cmap='RdYlGn', linewidths=0.5,
            vmin=-1, vmax=1, annot_kws={'size': 8})
plt.title('Feature Correlation Heatmap', fontsize=14)
plt.tight_layout()
plt.savefig('../data/plot5_correlation_heatmap.png', dpi=150)
plt.show()

print("\nCorrelation with placement (sorted high to low):")
corr_with_target = temp.corr()['placed'].drop('placed').sort_values(ascending=False)
print(corr_with_target.round(3).to_string())
```

```python
key_cols = ['cgpa', 'aptitude_skill_rating', 'coding_skill_rating',
            'attendance_percentage', 'internships_completed']

fig, axes = plt.subplots(1, len(key_cols), figsize=(20, 4))

for i, col in enumerate(key_cols):
    for status, color in [('Placed','#2ecc71'), ('Not Placed','#e74c3c')]:
        subset = df[df['placement_status'] == status][col]
        axes[i].hist(subset, bins=20, alpha=0.6,
                     color=color, label=status, edgecolor='white')
    axes[i].set_title(col, fontsize=11)
    axes[i].legend(fontsize=8)
    axes[i].set_xlabel('')

plt.suptitle('Distribution of Key Features by Placement', fontsize=13)
plt.tight_layout()
plt.savefig('../data/plot6_distributions.png', dpi=150)
plt.show()
```

```python
placed     = df[df['placement_status'] == 'Placed']
not_placed = df[df['placement_status'] == 'Not Placed']

print("=" * 50)
print("KEY INSIGHTS FROM EDA")
print("=" * 50)

insights = {
    'CGPA'                  : ('cgpa', ''),
    'Aptitude Skill Rating' : ('aptitude_skill_rating', ''),
    'Coding Skill Rating'   : ('coding_skill_rating', ''),
    'Attendance %'          : ('attendance_percentage', '%'),
    'Internships'           : ('internships_completed', ''),
    'Backlogs'              : ('backlogs', ''),
    'Study Hours/Day'       : ('study_hours_per_day', 'hrs'),
}

for label, (col, unit) in insights.items():
    p_mean  = placed[col].mean()
    np_mean = not_placed[col].mean()
    diff    = p_mean - np_mean
    arrow   = '↑' if diff > 0 else '↓'
    print(f"{label:<28} Placed: {p_mean:.2f}{unit}  |  "
          f"Not Placed: {np_mean:.2f}{unit}  |  "
          f"Diff: {arrow}{abs(diff):.2f}{unit}")
```
